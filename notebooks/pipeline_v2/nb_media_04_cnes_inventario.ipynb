# 04 — Inventário Bronze CNES (ST / LT)

**Reconstrução auxiliar:** inventário e diagnóstico dos controles Bronze de CNES; **não substitui** o notebook original de extração PySUS.

```python
from pathlib import Path
import json, pandas as pd
ROOT=Path('/lakehouse/default/Files/media')
CONTROLES=ROOT/'controle'/'cnes'
assert ROOT.exists(), 'Anexe o Lakehouse de origem ao notebook'
registros=[]
for arquivo in CONTROLES.glob('*.json'):
 try:
  x=json.loads(arquivo.read_text(encoding='utf-8'))
  registros.append({'grupo':x.get('grupo'),'uf':x.get('uf'),'ano':x.get('ano'),'mes':x.get('mes'),'status':x.get('status'),'destino':x.get('destino'),'linhas':x.get('linhas')})
 except Exception as exc: print('Controle invalido',arquivo.name,exc)
resumo=pd.DataFrame(registros)
print('Arquivos catalogados:',len(resumo))
if not resumo.empty: display(resumo.groupby(['grupo','ano','status'],dropna=False).size().reset_index(name='arquivos'))
```
